## 3. PyTorch 설치 확인

여기가 이 장에서 **가장 중요한 부분**이다. 세 가지를 본다.

1. PyTorch가 import 되는가
2. 버전 문자열 끝이 `+cu128`처럼 **CUDA 빌드**인가, 아니면 `+cpu`인가
3. `torch.cuda.is_available()`이 True인가

2장과 3장이 갈리는 지점을 정확히 이해해야 문제를 고칠 수 있다.

| 상황 | 버전 끝 | is_available | 원인 |
|---|---|---|---|
| 정상 | `+cu128` | True | — |
| CPU 빌드 설치됨 | `+cpu` | False | `--index-url` 누락 |
| 드라이버 문제 | `+cu128` | False | 드라이버가 오래됨 |

````python
try:
    import torch
    ok_torch = True
except ImportError:
    ok_torch = False
    print("[문제] PyTorch가 설치되지 않았습니다.")
    print("  1-4절의 설치 명령을 먼저 실행하세요.")

if ok_torch:
    print("=" * 55)
    print("PyTorch 설치 상태")
    print("=" * 55)
    print(f"버전          : {torch.__version__}")

    ver = torch.__version__
    is_cpu_build = "+cpu" in ver
    has_cu_tag = "+cu" in ver

    if has_cu_tag:
        print(f"빌드 종류      : CUDA 빌드 (컴파일 CUDA {torch.version.cuda})")
    elif is_cpu_build:
        print("빌드 종류      : CPU 전용 빌드")
    else:
        print("빌드 종류      : 태그 없음 (CPU 빌드일 가능성)")

    print(f"CUDA 사용 가능 : {torch.cuda.is_available()}")
    print("-" * 55)

    if torch.cuda.is_available():
        print("[OK] GPU를 사용할 수 있습니다.")
    elif is_cpu_build:
        print("[문제] CPU 전용 빌드가 설치되었습니다.")
        print("  해결:")
        print("    pip uninstall torch torchvision torchaudio -y")
        print("    pip cache purge")
        print("    pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu128")
    else:
        print("[문제] CUDA 빌드인데 GPU를 못 찾습니다.")
        print("  - NVIDIA 드라이버를 최신으로 업데이트하고 재부팅해 보세요.")
        print("  - 노트북이라면 Windows 그래픽 설정에서 Python을 '고성능'으로 지정해 보세요.")
    print()
    print("GPU가 없어도 이 책의 1~23장은 모두 실행됩니다.")
````

## 4. GPU 정보와 메모리 확인

GPU를 쓸 수 있다면 **메모리 용량**을 확인해 둔다. 이 값이 앞으로 다룰 수 있는 모델 크기를 결정한다.

이론편 22.5절에서 계산했듯, 파라미터 하나가 차지하는 공간은 정밀도에 따라 정해진다.

$$M = N_{params} \times \frac{b}{8}\ \text{bytes}$$

아래 셀은 이 계산을 실제 GPU 메모리에 대입해, **어느 규모까지 가능한지** 표로 보여준다.

````python
try:
    import torch
    HAS_TORCH = True
except ImportError:
    HAS_TORCH = False
    print("PyTorch 미설치 - 이 셀은 건너뜁니다.")

if HAS_TORCH and torch.cuda.is_available():
    idx = torch.cuda.current_device()
    props = torch.cuda.get_device_properties(idx)
    total_gb = props.total_memory / 1024**3

    print("=" * 55)
    print("GPU 정보")
    print("=" * 55)
    print(f"이름          : {props.name}")
    print(f"전체 메모리    : {total_gb:.1f} GB")
    print(f"멀티프로세서   : {props.multi_processor_count}개")
    print(f"Compute 능력   : {props.major}.{props.minor}")

    alloc = torch.cuda.memory_allocated(idx) / 1024**3
    reserved = torch.cuda.memory_reserved(idx) / 1024**3
    print(f"현재 할당      : {alloc:.2f} GB")
    print(f"현재 예약      : {reserved:.2f} GB")

    # 실사용 가능 메모리는 전체의 약 85% 정도로 잡는 것이 안전
    usable = total_gb * 0.85
    print("-" * 55)
    print(f"실사용 가능(추정) : 약 {usable:.1f} GB")
    print()
    print("이 GPU로 가능한 작업 (이론편 22.5절 계산 기준)")
    print("-" * 55)
    print(f"{'모델 크기':<10}{'FP16 추론':<12}{'INT4 추론':<12}{'QLoRA 학습':<12}")
    print("-" * 55)
    for n, name in [(0.5e9, "0.5B"), (1.5e9, "1.5B"), (3e9, "3B"), (7e9, "7B")]:
        fp16 = n * 2 / 1024**3
        int4 = n * 0.5 / 1024**3
        qlora = int4 * 1.6 + 1.0          # 활성값·옵티마이저 여유 개략
        m = lambda v: "가능" if v < usable else "어려움"
        print(f"{name:<10}{m(fp16):<12}{m(int4):<12}{m(qlora):<12}")
    print("-" * 55)
    print("※ 실제로는 배치 크기·시퀀스 길이에 따라 달라집니다.")
elif HAS_TORCH:
    print("GPU를 사용할 수 없어 이 셀은 건너뜁니다.")
    print("CPU로도 1~23장은 모두 실행할 수 있습니다.")
````

## 5. 기본 패키지 확인

앞으로 계속 쓸 패키지들이 설치되었는지 한 번에 확인한다.
없는 것이 있으면 아래 출력에 설치 명령이 함께 표시된다.

````python
import importlib

packages = [
    ("numpy",        "수치 계산 (2장부터)"),
    ("matplotlib",   "시각화"),
    ("pandas",       "표 데이터 (8장부터)"),
    ("scipy",        "과학 계산"),
    ("sklearn",      "머신러닝 (8장부터)"),
    ("tqdm",         "진행 표시"),
    ("IPython",      "주피터 노트북 환경"),
]

print("=" * 55)
print("패키지 설치 확인")
print("=" * 55)

missing = []
for name, desc in packages:
    try:
        mod = importlib.import_module(name)
        ver = getattr(mod, "__version__", "버전 정보 없음")
        print(f"[OK]  {name:<14}{ver:<12}{desc}")
    except ImportError:
        print(f"[없음] {name:<14}{'':<12}{desc}")
        missing.append(name)

print("-" * 55)
if missing:
    install_names = ["scikit-learn" if m == "sklearn" else m for m in missing]
    print("설치가 필요합니다. 터미널에서 실행하세요:")
    print(f"  pip install {' '.join(install_names)}")
else:
    print("[OK] 필요한 패키지가 모두 설치되어 있습니다.")
````

## 6. 실제 연산 테스트

설치 확인만으로는 부족하다. **실제로 계산이 되는지** 확인해야 한다.

여기서는 두 가지를 본다.

1. CPU에서 행렬 곱이 되는가
2. GPU가 있다면, GPU에서도 되는가 — 그리고 얼마나 빠른가

이론편 4.2절에서 손으로 계산했던 행렬 곱을 그대로 써서 값이 맞는지도 함께 확인한다.

````python
import time

try:
    import torch
    HAS_TORCH = True
except ImportError:
    HAS_TORCH = False

print("=" * 55)
print("연산 테스트")
print("=" * 55)

# --- 5-1. 정확성: 이론편 4.2절 값과 대조 ---
print("[정확성] 이론편 4.2절 예제  A @ x")
if HAS_TORCH:
    A = torch.tensor([[2.0, -1.0], [1.0, 3.0]])
    x = torch.tensor([3.0, 2.0])
    y = (A @ x).tolist()
else:
    import numpy as np
    A = np.array([[2.0, -1.0], [1.0, 3.0]])
    x = np.array([3.0, 2.0])
    y = (A @ x).tolist()
    print("  (PyTorch 미설치 - NumPy로 확인)")

print(f"  결과   = {y}")
print(f"  기대값 = [4.0, 9.0]")
assert abs(y[0] - 4.0) < 1e-6 and abs(y[1] - 9.0) < 1e-6, "계산 결과가 다릅니다"
print("  [OK] 이론편 손계산과 일치")

# --- 5-2. 속도 비교 ---
if HAS_TORCH:
    print()
    print("[속도] 2000x2000 행렬 곱")
    size = 2000
    a_cpu = torch.randn(size, size)
    b_cpu = torch.randn(size, size)

    t0 = time.time()
    _ = a_cpu @ b_cpu
    cpu_time = time.time() - t0
    print(f"  CPU : {cpu_time*1000:8.1f} ms")

    if torch.cuda.is_available():
        a_gpu, b_gpu = a_cpu.cuda(), b_cpu.cuda()

        _ = a_gpu @ b_gpu          # 워밍업 (첫 실행은 초기화 비용 포함)
        torch.cuda.synchronize()

        t0 = time.time()
        _ = a_gpu @ b_gpu
        torch.cuda.synchronize()   # GPU는 비동기이므로 동기화 후 측정
        gpu_time = time.time() - t0

        print(f"  GPU : {gpu_time*1000:8.1f} ms")
        print(f"  배율 : {cpu_time/gpu_time:8.1f}배 빠름")
        print()
        print("  [OK] GPU 연산 정상")
    else:
        print("  GPU : 사용 불가 (건너뜀)")
else:
    print()
    print("PyTorch 설치 후 속도 비교를 확인할 수 있습니다.")
````

### 왜 `torch.cuda.synchronize()`가 필요한가

GPU 연산은 **비동기**로 동작한다. Python이 `a_gpu @ b_gpu`를 호출하면 GPU에 작업을 맡기고
결과를 기다리지 않은 채 다음 줄로 넘어간다.

동기화 없이 시간을 재면 "작업을 맡기는 데 걸린 시간"만 측정되어, GPU가 터무니없이 빨라 보인다.
`synchronize()`는 GPU가 실제로 계산을 끝낼 때까지 기다리게 한다.

앞으로 GPU 성능을 잴 때마다 이 함수가 필요하다.

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import platform

_c = {"Windows": ["Malgun Gothic"], "Darwin": ["AppleGothic"],
      "Linux": ["NanumGothic", "Noto Sans CJK KR", "Noto Sans CJK JP"]}
_a = {f.name for f in fm.fontManager.ttflist}
for _n in _c.get(platform.system(), []):
    if _n in _a:
        plt.rcParams["font.family"] = _n
        break
plt.rcParams["axes.unicode_minus"] = False

import torch

if torch.cuda.is_available():
    vram = torch.cuda.get_device_properties(0).total_memory / 1024**3
else:
    vram = 0.0
usable = vram * 0.85

fig, axes = plt.subplots(1, 2, figsize=(13, 4.2))

# --- 왼쪽: 실습별 필요 메모리 ---
ax = axes[0]
tasks = [
    ("1~10장 기초", 0.0),
    ("11~18장 딥러닝", 0.0),
    ("19~23장 LLM·RAG", 0.0),
    ("24장 SFT (소형)", 2.0),
    ("25장 QLoRA 3B", 5.0),
    ("25장 QLoRA 7B", 6.5),
    ("전체 파인튜닝 1B", 9.0),
]
names = [t[0] for t in tasks]
needs = [t[1] for t in tasks]
colors = []
for n in needs:
    if n == 0.0:
        colors.append("#0D9488")
    elif vram > 0 and usable >= n:
        colors.append("#1E40AF")
    else:
        colors.append("#DC2626")

ax.barh(range(len(names)), [max(n, 0.4) for n in needs], color=colors)
if vram > 0:
    ax.axvline(usable, color="#EA580C", linestyle="--", linewidth=2)
    ax.text(usable + 0.2, 0.2, f"사용 가능\n{usable:.1f}GB",
            fontsize=8, color="#EA580C")
ax.set_yticks(range(len(names)))
ax.set_yticklabels(names, fontsize=8)
ax.set_xlabel("필요 VRAM (GB)")
ax.set_title("초록 = CPU 로 가능 / 파랑 = 내 GPU 로 가능 / 빨강 = 어려움")
ax.invert_yaxis()
ax.grid(axis="x", alpha=0.3)

# --- 오른쪽: 모델 크기별 메모리 (이론편 22.5절) ---
ax = axes[1]
sizes = np.array([0.1, 0.5, 1.5, 3, 7, 13])
fp16 = sizes * 2
int4 = sizes * 0.5
train = sizes * 8

ax.plot(sizes, fp16, marker="o", linewidth=2, color="#0D9488", label="추론 FP16")
ax.plot(sizes, int4, marker="s", linewidth=2, color="#1E40AF", label="추론 INT4")
ax.plot(sizes, train, marker="^", linewidth=2, color="#DC2626", label="전체 학습")
if vram > 0:
    ax.axhline(usable, color="#EA580C", linestyle="--", linewidth=2)
    ax.text(0.15, usable * 1.15, f"내 GPU {usable:.1f}GB",
            fontsize=8, color="#EA580C")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("모델 크기 (B, 로그)")
ax.set_ylabel("필요 메모리 GB (로그)")
ax.set_title("모델 크기와 메모리")
ax.legend(fontsize=8)
ax.grid(alpha=0.3, which="both")

plt.tight_layout()
plt.show()

if vram == 0:
    print("GPU 가 없어도 1~31장을 CPU 로 실행할 수 있다.")
    print("  각 실습은 CPU 기준으로 모델 크기와 학습량을 조정했다.")
else:
    print(f"VRAM {vram:.1f}GB — 사용 가능 약 {usable:.1f}GB")
    if usable >= 6.5:
        print("  이 책의 모든 실습을 진행할 수 있다.")
    elif usable >= 4:
        print("  대부분 가능하나 7B급 QLoRA 는 배치를 줄여야 한다.")
    else:
        print("  파인튜닝 실습은 더 작은 모델로 대체하는 편이 낫다.")
print()
print("오른쪽 그래프의 요지 (이론편 22.5절)")
print("  같은 모델도 정밀도에 따라 필요 메모리가 4배까지 차이난다.")
print("  학습은 추론의 4배가 든다 — 가중치 + 그래디언트 + 옵티마이저 상태")
print("  → 32장에서 이 문제를 푸는 방법을 다룬다.")
````
