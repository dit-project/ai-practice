````python
import numpy as np

print("=" * 70)
print("핵심 값 재확인 — 지금 다시 계산해 본다")
print("=" * 70)

# 이론편 4.2절
A = np.array([[2.0, -1.0], [1.0, 3.0]])
x = np.array([3.0, 2.0])
assert np.allclose(A @ x, [4.0, 9.0])
print(f"  [이론편 4.2] A @ x = {A @ x}  OK")

# 이론편 4.3절
A2 = np.array([[3.0, 1.0], [0.0, 2.0]])
w = np.linalg.eigvals(A2)
assert np.allclose(sorted(w, reverse=True), [3.0, 2.0])
print(f"  [이론편 4.3] 고유값 = {sorted(w, reverse=True)}  OK")

# 이론편 10.3절
sigmoid = lambda z: 1 / (1 + np.exp(-z))
d = sigmoid(0) * (1 - sigmoid(0))
assert abs(d - 0.25) < 1e-6
print(f"  [이론편 10.3] 시그모이드 도함수 최대 = {d}  OK")

# 이론편 18.3절 (20장의 핵심)
def softmax(v):
    e = np.exp(v - v.max())
    return e / e.sum()

Q = np.array([[1.0, 0.0], [0.0, 1.0], [1.0, 1.0]])
scores = (Q @ Q.T) / np.sqrt(2)
alpha = softmax(scores[0])
assert np.allclose(alpha, [0.401, 0.198, 0.401], atol=0.001)
print(f"  [이론편 18.3] Attention α = {alpha.round(3)}  OK")

# 이론편 22.4절 (32장)
d_model, r = 4096, 8
ratio = (2 * d_model * r) / (d_model * d_model) * 100
assert abs(ratio - 0.39) < 0.01
print(f"  [이론편 22.4] LoRA 비율 = {ratio:.2f}%  OK")

# 이론편 23.3절 (34장)
import math
loss = -math.log(1 / (1 + math.exp(-0.1 * 1.0)))
assert abs(loss - 0.644) < 0.001
print(f"  [이론편 23.3] DPO 손실 = {loss:.4f}  OK")

print("-" * 70)
print("[OK] 핵심 값 6개 모두 재확인")
````

## 3. 재사용 가능한 도구 모음

여러 장에서 만든 함수들을 **하나의 모듈로 정리한다.**
43~44번 프로젝트에서 이것을 가져다 쓴다.

````python
from pathlib import Path

root = Path.cwd()
if root.name.startswith("part"):
    root = root.parent

utils_dir = root / "utils"
utils_dir.mkdir(exist_ok=True)

verify_code = '''"""검증 도구 — 05장, 11장에서 만든 것"""
import numpy as np


def verify_gradient(loss_fn, grad_fn, params, h=1e-5, tol=1e-4, verbose=True):
    """수치 미분으로 그래디언트 구현이 맞는지 확인 (05장 5절)"""
    analytic = np.asarray(grad_fn(params), dtype=float)
    numeric = np.zeros_like(analytic)
    params = np.asarray(params, dtype=float)

    for i in range(len(params)):
        p_plus = params.copy();  p_plus[i] += h
        p_minus = params.copy(); p_minus[i] -= h
        numeric[i] = (loss_fn(p_plus) - loss_fn(p_minus)) / (2 * h)

    diff = np.abs(analytic - numeric).max()
    passed = diff < tol

    if verbose:
        print(f"  수식 유도 : {analytic.round(6)}")
        print(f"  수치 미분 : {numeric.round(6)}")
        print(f"  최대 차이 : {diff:.2e}  -> {'통과' if passed else '실패'}")
    return passed


def verify_against_book(actual, expected, label, tol=1e-3, verbose=True):
    """이론편 손계산 값과 대조 (05장 5절)"""
    actual = np.atleast_1d(np.asarray(actual, dtype=float))
    expected = np.atleast_1d(np.asarray(expected, dtype=float))
    passed = np.allclose(actual, expected, atol=tol)

    if verbose:
        mark = "통과" if passed else "실패"
        print(f"  [{mark}] {label}")
        print(f"         계산값 {actual.round(4)}  /  이론편 값 {expected.round(4)}")
    return passed


def softmax(x, axis=-1):
    """수치적으로 안정된 소프트맥스 (20장 4절)"""
    x_shifted = x - x.max(axis=axis, keepdims=True)
    e = np.exp(x_shifted)
    return e / e.sum(axis=axis, keepdims=True)


def cosine_similarity(a, b):
    """코사인 유사도 (27장 2절)"""
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))
'''

(utils_dir / "verify.py").write_text(verify_code, encoding="utf-8")
print(f"생성: {utils_dir / 'verify.py'}")
````

## 4. 부품별 점검

프로젝트에 들어가기 전에 **필요한 부품들이 준비됐는지** 확인한다.

````python
import importlib

print("=" * 78)
print("패키지 점검")
print("=" * 78)

groups = {
    "기초 (01~12장)": [
        ("numpy", "수치 계산", True),
        ("matplotlib", "시각화", True),
        ("sklearn", "머신러닝", True),
        ("torch", "딥러닝", True),
    ],
    "LLM (23~25장)": [
        ("transformers", "Hugging Face", True),
        ("openai", "API 연동", False),
        ("dotenv", "환경 변수", False),
    ],
    "검색·RAG (27~28장)": [
        ("sentence_transformers", "문장 임베딩", True),
        ("chromadb", "벡터 DB", False),
    ],
    "파인튜닝 (31~34장)": [
        ("peft", "LoRA", False),
    ],
    "기타": [
        ("PIL", "이미지 (36장)", False),
        ("gymnasium", "강화학습 (17장)", False),
    ],
}

status = {}
for group, items in groups.items():
    print(f"\n[{group}]")
    for name, desc, essential in items:
        try:
            mod = importlib.import_module(name)
            ver = getattr(mod, "__version__", "설치됨")
            print(f"  [OK]   {name:<24}{ver:<14}{desc}")
            status[name] = True
        except ImportError:
            mark = "필수" if essential else "선택"
            print(f"  [{mark}] {name:<24}{'':<14}{desc}")
            status[name] = False

print()
print("-" * 78)
n_ok = sum(status.values())
print(f"설치됨: {n_ok}/{len(status)}")

missing_essential = [n for g in groups.values() for n, d, e in g
                     if e and not status.get(n)]
if missing_essential:
    print(f"\n[주의] 필수 패키지 미설치: {missing_essential}")
    print("  pip install -r requirements.txt 를 실행하세요.")
else:
    print("\n필수 패키지가 모두 준비되었습니다.")
````
